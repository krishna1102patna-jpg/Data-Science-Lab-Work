```python
#Assign variables:

TP = 50
FP = 8
TN = 32
FN = 10
```

```python
#Calculate total:

total = TP + FP + TN + FN
```

```python
#Calculate metrics:

accuracy = (TP + TN) / total
precision = TP / (TP + FP)
recall = TP / (TP + FN)
f1_score = 2 * (precision * recall) / (precision + recall)
```

```python
#Print results with labels:

print("Total Predictions:", total)
print("Accuracy:", accuracy)
print("Precision:", precision)
print("Recall:", recall)
print("F1 Score:", f1_score)
```

```python
errors = [2, 4, 6, 8, 10]

total_error = errors[0] + errors[1] + errors[2] + errors[3] + errors[4]
mse = total_error / 5

print("Mean Squared Error:", mse)
```

```python
e2 = 9
MSE = 3
p = 3
h = 0.2
n = 25

cook_d = (e2 / (p * MSE)) * (h / (1 - h)**2)
threshold = 4 / n

print("Cook's Distance:", cook_d)
print("Threshold:", threshold)
```

```python
#Decision:

if cook_d > threshold:
    print("Observation is influential")
else:
    print("Observation is not influential")
```
